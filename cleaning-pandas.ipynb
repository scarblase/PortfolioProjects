```python
import pandas as pd
```

```python
data = {
    'CustomerID': [101, 102, 103, 104, None, 106, 107],
    'CustomerName': ['Alice', 'BOB', 'Charlie', 'DAVE', None, 'eve', 'eve'],
    'Age': [25, None, 35, 40, 28, 30, 30],
    'OrderDate': ['2023/01/15', '2023-02-20', 'March 10, 2023', None, '2023/04/10', '2023-05-05', 'InvalidDate'],
    'OrderAmount': [200, 300, None, 400, 250, None, 150],
    'Region': ['North', 'South', 'East', None, 'West', 'South', 'East'],
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', None, 'Clothing'],
    'FeedbackScore': [5, 3, None, 4, 2, None, -1],
    'ExtraColumn': [None, None, None, None, None, None, None]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.drop(columns=['ExtraColumn'])

df['CustomerID'] = df['CustomerID'].fillna(method='ffill').astype(int)  
df['CustomerName'] = df['CustomerName'].fillna('Unknown').str.title() 
df['Age'] = df['Age'].fillna(round(df['Age'].mean(), 2))
df['OrderAmount'] = df['OrderAmount'].fillna(0) 
df['Region'] = df['Region'].fillna('Unknown')
df['Category'] = df['Category'].fillna('Miscellaneous') 
df['FeedbackScore'] = df['FeedbackScore'].fillna(0)

df
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors='coerce')

df['OrderDate'] = df['OrderDate'].fillna(pd.Timestamp('2023-01-01'))

print(df)
```

```python
df = df.drop_duplicates(subset=['CustomerID', 'OrderDate'])

df = df[df['FeedbackScore'] >= 0]  
```

```python
df['OrderYear'] = df['OrderDate'].dt.year  
df['OrderMonth'] = df['OrderDate'].dt.month
```

```python
df['AgeGroup'] = pd.cut(df['Age'], bins=[0, 18, 35, 50, 100], labels=['Teen', 'Young Adult', 'Adult', 'Senior'])  
```
